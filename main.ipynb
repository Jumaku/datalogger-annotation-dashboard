# Manual Supercooling Event Dashboard

Run this after `data_dict` has been loaded and trimmed. Use it to annotate each experiment/channel with the manually selected spike start, spike maximum, and spike end.


```python
# Manual supercooling event dashboard
# Run this after data_dict has been loaded and trimmed.

import os
import pandas as pd
import plotly.graph_objects as go
from IPython.display import display, clear_output

try:
    import ipywidgets as widgets
except ImportError as exc:
    raise ImportError(
        "This dashboard requires ipywidgets. Install it with: "
        "python -m pip install ipywidgets"
    ) from exc

if "data_dict" not in globals() or not data_dict:
    raise NameError(
        "Load data_dict before running this dashboard, for example with "
        "setup_data_dict(TARGET_EXP)."
    )

EVENT_LABELS = {
    "start": "Spike start / supercooling",
    "maximum": "Spike maximum",
    "end": "Spike end / decline",
}

manual_event_columns = [
    "experiment_id",
    "channel",
    "start_time_min",
    "start_temp_c",
    "maximum_time_min",
    "maximum_temp_c",
    "end_time_min",
    "end_temp_c",
    "notes",
]
manual_events_path = os.path.join(APP_DIR["output"], "manual_supercooling_events.csv")

if "manual_supercooling_events" not in globals():
    if os.path.exists(manual_events_path):
        manual_supercooling_events = pd.read_csv(manual_events_path)
    else:
        manual_supercooling_events = pd.DataFrame(columns=manual_event_columns)

manual_supercooling_events = manual_supercooling_events.reindex(
    columns=manual_event_columns
)


def _tc_columns(df):
    if "get_tc_columns" in globals():
        return get_tc_columns(df)
    return sorted(
        [col for col in df.columns if str(col).strip().upper().startswith("TC")],
        key=lambda col: int("".join(filter(str.isdigit, str(col))) or 9999),
    )


def _current_df():
    return data_dict[experiment_picker.value]


def _current_channel():
    return channel_picker.value


def _time_bounds(df):
    time_min = float(df["Time"].min() / 60)
    time_max = float(df["Time"].max() / 60)
    step = max((time_max - time_min) / 1000, 0.001)
    return time_min, time_max, step


def _nearest_temp(df, channel, time_min):
    target_seconds = float(time_min) * 60
    idx = (df["Time"] - target_seconds).abs().idxmin()
    return float(df.loc[idx, channel]), float(df.loc[idx, "Time"] / 60)


def _set_time_widget_bounds():
    df = _current_df()
    time_min, time_max, step = _time_bounds(df)
    for widget in time_widgets.values():
        widget.min = time_min
        widget.max = time_max
        widget.step = step
        widget.value = min(max(widget.value, time_min), time_max)


def _load_saved_values():
    global manual_supercooling_events

    mask = (
        (manual_supercooling_events["experiment_id"] == experiment_picker.value)
        & (manual_supercooling_events["channel"] == channel_picker.value)
    )

    if not mask.any():
        _snap_all_temperatures()
        notes.value = ""
        return

    row = manual_supercooling_events.loc[mask].iloc[-1]
    for event in EVENT_LABELS:
        saved_time = row.get(f"{event}_time_min")
        saved_temp = row.get(f"{event}_temp_c")
        if pd.notna(saved_time):
            time_widgets[event].value = float(saved_time)
        if pd.notna(saved_temp):
            temp_widgets[event].value = float(saved_temp)
    notes.value = "" if pd.isna(row.get("notes")) else str(row.get("notes"))


def _snap_temperature(event):
    df = _current_df()
    channel = _current_channel()
    temp, snapped_time = _nearest_temp(df, channel, time_widgets[event].value)
    time_widgets[event].value = snapped_time
    temp_widgets[event].value = temp


def _snap_all_temperatures(*_):
    for event in EVENT_LABELS:
        _snap_temperature(event)


def _event_row():
    return {
        "experiment_id": experiment_picker.value,
        "channel": channel_picker.value,
        "start_time_min": time_widgets["start"].value,
        "start_temp_c": temp_widgets["start"].value,
        "maximum_time_min": time_widgets["maximum"].value,
        "maximum_temp_c": temp_widgets["maximum"].value,
        "end_time_min": time_widgets["end"].value,
        "end_temp_c": temp_widgets["end"].value,
        "notes": notes.value,
    }


def _save_event(*_):
    global manual_supercooling_events

    row = _event_row()
    key_mask = (
        (manual_supercooling_events["experiment_id"] == row["experiment_id"])
        & (manual_supercooling_events["channel"] == row["channel"])
    )

    if key_mask.any():
        for col in manual_event_columns:
            manual_supercooling_events.loc[key_mask, col] = row[col]
    else:
        manual_supercooling_events = pd.concat(
            [manual_supercooling_events, pd.DataFrame([row])],
            ignore_index=True,
        )

    status.value = (
        f"Saved {row['experiment_id']} {row['channel']} "
        f"({len(manual_supercooling_events)} annotated channel(s))."
    )
    _render_table()
    _render_plot()


def _export_events(*_):
    if manual_supercooling_events.empty:
        status.value = "No manual annotations to export yet."
        return

    os.makedirs(os.path.dirname(manual_events_path), exist_ok=True)
    manual_supercooling_events.to_csv(manual_events_path, index=False)
    status.value = f"Exported {len(manual_supercooling_events)} row(s) to {manual_events_path}"


def _change_channel(offset):
    channels = list(channel_picker.options)
    idx = channels.index(channel_picker.value)
    channel_picker.value = channels[(idx + offset) % len(channels)]


def _render_plot(*_):
    df = _current_df()
    channel = _current_channel()
    label = channel_layout.get(experiment_picker.value, {}).get(channel, channel)

    fig = go.Figure()
    fig.add_trace(
        go.Scatter(
            x=df["Time"] / 60,
            y=df[channel],
            mode="lines",
            name=f"{channel} - {label}",
        )
    )

    marker_styles = {
        "start": {"color": "royalblue", "symbol": "diamond"},
        "maximum": {"color": "firebrick", "symbol": "circle"},
        "end": {"color": "seagreen", "symbol": "square"},
    }

    for event, event_label in EVENT_LABELS.items():
        event_time = time_widgets[event].value
        event_temp = temp_widgets[event].value
        fig.add_vline(
            x=event_time,
            line_width=1,
            line_dash="dash",
            line_color=marker_styles[event]["color"],
        )
        fig.add_trace(
            go.Scatter(
                x=[event_time],
                y=[event_temp],
                mode="markers+text",
                name=event_label,
                marker={
                    "size": 11,
                    "color": marker_styles[event]["color"],
                    "symbol": marker_styles[event]["symbol"],
                },
                text=[event_label],
                textposition="top center",
            )
        )

    fig.update_layout(
        title=f"Manual supercooling annotation: {experiment_picker.value} {channel}",
        xaxis_title="Time [min]",
        yaxis_title="Temperature [deg C]",
        template="plotly_white",
        hovermode="x unified",
        width=1100,
        height=520,
        margin={"l": 70, "r": 30, "t": 70, "b": 60},
    )

    with plot_output:
        clear_output(wait=True)
        fig.show()


def _render_table():
    with table_output:
        clear_output(wait=True)
        if manual_supercooling_events.empty:
            display(pd.DataFrame(columns=manual_event_columns))
        else:
            display(
                manual_supercooling_events.sort_values(
                    ["experiment_id", "channel"]
                ).reset_index(drop=True)
            )


def _on_experiment_change(change):
    if change["name"] != "value":
        return
    channels = _tc_columns(_current_df())
    channel_picker.options = channels
    channel_picker.value = channels[0]
    _set_time_widget_bounds()
    _load_saved_values()
    _render_plot()
    _render_table()


def _on_channel_change(change):
    if change["name"] != "value":
        return
    _load_saved_values()
    _render_plot()
    _render_table()


experiment_picker = widgets.Dropdown(
    options=sorted(data_dict.keys()),
    description="Experiment",
    layout=widgets.Layout(width="260px"),
)
channel_picker = widgets.Dropdown(
    options=_tc_columns(data_dict[sorted(data_dict.keys())[0]]),
    description="Channel",
    layout=widgets.Layout(width="220px"),
)

initial_df = data_dict[experiment_picker.value]
time_min, time_max, time_step = _time_bounds(initial_df)
initial_start = time_min + (time_max - time_min) * 0.33
initial_maximum = time_min + (time_max - time_min) * 0.36
initial_end = time_min + (time_max - time_min) * 0.40

time_widgets = {
    "start": widgets.BoundedFloatText(
        value=initial_start,
        min=time_min,
        max=time_max,
        step=time_step,
        description="Start min",
        layout=widgets.Layout(width="190px"),
    ),
    "maximum": widgets.BoundedFloatText(
        value=initial_maximum,
        min=time_min,
        max=time_max,
        step=time_step,
        description="Max min",
        layout=widgets.Layout(width="190px"),
    ),
    "end": widgets.BoundedFloatText(
        value=initial_end,
        min=time_min,
        max=time_max,
        step=time_step,
        description="End min",
        layout=widgets.Layout(width="190px"),
    ),
}

temp_widgets = {
    event: widgets.FloatText(
        value=0.0,
        description="Temp C",
        layout=widgets.Layout(width="170px"),
    )
    for event in EVENT_LABELS
}

notes = widgets.Text(
    value="",
    placeholder="optional note",
    description="Notes",
    layout=widgets.Layout(width="520px"),
)
status = widgets.HTML(value="")
plot_output = widgets.Output()
table_output = widgets.Output()

snap_buttons = {
    event: widgets.Button(description=f"Snap {event}", button_style="")
    for event in EVENT_LABELS
}
snap_all_button = widgets.Button(description="Snap all temps", button_style="info")
save_button = widgets.Button(description="Save/update row", button_style="success")
export_button = widgets.Button(description="Export CSV", button_style="primary")
prev_channel_button = widgets.Button(description="Previous TC")
next_channel_button = widgets.Button(description="Next TC")

for event, button in snap_buttons.items():
    button.on_click(lambda _, event=event: (_snap_temperature(event), _render_plot()))
snap_all_button.on_click(lambda _: (_snap_all_temperatures(), _render_plot()))
save_button.on_click(_save_event)
export_button.on_click(_export_events)
prev_channel_button.on_click(lambda _: _change_channel(-1))
next_channel_button.on_click(lambda _: _change_channel(1))

experiment_picker.observe(_on_experiment_change, names="value")
channel_picker.observe(_on_channel_change, names="value")
for widget in list(time_widgets.values()) + list(temp_widgets.values()):
    widget.observe(_render_plot, names="value")

# Build event rows explicitly so each event has a label, time, temperature, and snap button.
event_rows = []
for event, event_label in EVENT_LABELS.items():
    event_rows.extend(
        [
            widgets.HTML(f"<b>{event_label}</b>"),
            time_widgets[event],
            temp_widgets[event],
            snap_buttons[event],
        ]
    )

event_grid = widgets.GridBox(
    children=event_rows,
    layout=widgets.Layout(
        grid_template_columns="220px 200px 180px 130px",
        grid_gap="6px 10px",
        align_items="center",
    ),
)

dashboard = widgets.VBox(
    [
        widgets.HBox([experiment_picker, channel_picker, prev_channel_button, next_channel_button]),
        event_grid,
        widgets.HBox([snap_all_button, save_button, export_button]),
        notes,
        status,
        plot_output,
        table_output,
    ]
)

_snap_all_temperatures()
_render_plot()
_render_table()
display(dashboard)
```
