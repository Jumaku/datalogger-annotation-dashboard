# Click Dashboard For Manual Supercooling Events

This notebook starts a local dashboard server. It does not use `ipywidgets`, so it avoids the `VBox(...)` rendering problem.

The dashboard lets you:

- choose an experiment from `data/data_raw/raw_data.h5`
- choose a thermocouple channel
- click the trace three times: first supercooling temperature, high spike temperature, event end
- mark a channel as having no clear cooling event
- update and persist a dataframe below the plot

Annotations are saved to `output/manual_supercooling_events.csv`.


```python
from IPython.display import HTML, IFrame, display
from manual_dashboard_server import start_dashboard_server

DASHBOARD_PORT = 8765

try:
    manual_dashboard_server.shutdown()
    manual_dashboard_server.server_close()
except NameError:
    pass
except Exception:
    pass

manual_dashboard_server, manual_dashboard_thread, manual_dashboard_state = start_dashboard_server(
    port=DASHBOARD_PORT,
)

dashboard_url = f"http://127.0.0.1:{DASHBOARD_PORT}/"
print(f"Dashboard running at: {dashboard_url}")
print(f"Loaded experiments: {len(manual_dashboard_state.data_dict)}")
print(f"Annotations CSV: {manual_dashboard_state.annotations_path}")

display(HTML(f'<p><a href="{dashboard_url}" target="_blank">Open manual supercooling click dashboard</a></p>'))
display(IFrame(src=dashboard_url, width="100%", height=980))
```

## Stop The Dashboard

Run this cell when you are done or before restarting the dashboard on the same port.


```python
manual_dashboard_server.shutdown()
manual_dashboard_server.server_close()
print("Dashboard stopped")
```
